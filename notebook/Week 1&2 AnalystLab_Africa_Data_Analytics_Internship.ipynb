**Task 4: Data Visualization**

```python
# Import the Matplotlib library for creating visualizations
import matplotlib.pyplot as plt
```

**📊 Online Retail Dataset (5 Visualizations)**

**Visualization 1: Top 10 Countries by Transactions**

```python
# Display the top 10 countries with the highest number of transactions
df["Country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Countries by Number of Transactions")
plt.xlabel("Country")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()
```

**Visualization 2: Top 10 Best-Selling Products**

```python
# Display the top 10 most frequently sold products
df["Description"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Best-Selling Products")
plt.xlabel("Product")
plt.ylabel("Number of Sales")
plt.xticks(rotation=90)
plt.show()
```

**Visualization 3: Quantity Distribution**

```python
# Display the distribution of product quantities sold
plt.hist(df["Quantity"], bins=30)

plt.title("Distribution of Product Quantities")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()
```

**Visualization 4: Unit Price Distribution**

```python
# Display the distribution of product prices
plt.hist(df["UnitPrice"], bins=30)

plt.title("Distribution of Unit Prices")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.show()
```

**Visualization 5: Top 10 Countries by Quantity Sold**

```python
# Display the top 10 countries by total quantity sold
df.groupby("Country")["Quantity"].sum().sort_values(ascending=False).head(10).plot(kind="bar")

plt.title("Top 10 Countries by Quantity Sold")
plt.xlabel("Country")
plt.ylabel("Total Quantity")
plt.xticks(rotation=45)
plt.show()
```

**Netflix Dataset (5 Visualizations)**

**Visualization 1: Movies vs TV Shows**

```python
# Display the distribution of Movies and TV Shows
netflix_df["type"].value_counts().plot(kind="bar")

plt.title("Distribution of Netflix Content")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

**Visualization 2: Top 10 Ratings**

```python
# Display the top 10 most common content ratings
netflix_df["rating"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

**Visualization 3: Top 10 Countries**

```python
# Display the top 10 countries producing Netflix content
netflix_df["country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=90)
plt.show()
```

**Visualization 4: Release Year Distribution**

```python
# Display the distribution of release years
plt.hist(netflix_df["release_year"], bins=30)

plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

**Visualization 5: Top 10 Genres**

```python
# Display the top 10 most common genres
netflix_df["listed_in"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=90)
plt.show()
```
